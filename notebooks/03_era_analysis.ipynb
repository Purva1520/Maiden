# Era Analysis

How the scoring environment shifts across time (era validation, not game UI).

```python
import pandas as pd, matplotlib.pyplot as plt
eb = pd.read_parquet('../data/processed/era_baselines.parquet')
sr = eb[(eb.metric=='bat_strike_rate')&(eb.population=='batting')]
sr.plot.bar(x='era_id', y='median', legend=False, title='Median batting SR by era'); plt.tight_layout()
```
